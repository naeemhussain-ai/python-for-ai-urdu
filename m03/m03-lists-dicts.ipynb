## List of dicts + for loop

```python
expenses = [{"item": "Chai", "amount": 80}, {"item": "Biryani", "amount": 450}, {"item": "Petrol", "amount": 3000}]
expenses[1]["item"]
```

## Project: Expense Tracker
Khali Enter = kharche khatam.

```python
print("Expense Tracker shuru! (khali Enter = band)")
expenses = []

while True:
    item = input("Kharcha kis cheez ka? ").strip().title()
    if item == "":
        break

    amount = int(input("Kitne rupay? "))
    category = input("Category? (khana/safar/bill) ").strip().lower()

    expense = {"item": item, "amount": amount, "category": category}
    expenses.append(expense)
    print(f"Add hua: {item}, Rs {amount:,} [{category.upper()}]")

print("--- Aap ke kharche ---")
total = 0
for expense in expenses:
    print(f"{expense['item']}: Rs {expense['amount']:,}")
    total = total + expense["amount"]

print(f"Total: Rs {total:,}")
if total > 5000:
    print("Bhai, aaj kharcha zyada ho gaya!")
```
