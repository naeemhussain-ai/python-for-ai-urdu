## return vs print

```python
def add_print(a, b):
    print(a + b)

def add_return(a, b):
    return a + b
```

## Type hints, default aur keyword arguments

```python
def format_rs(amount: int, currency: str = "Rs") -> str:
    return f"{currency} {amount:,}"

format_rs(1500)
```

## Project: Expense Tracker
Pehle **functions** wala cell chalao, phir **program** wala.

```python
# functions
def format_rs(amount: int) -> str:
    """Raqam ko 'Rs 1,500' ki shakal mein do."""
    return f"Rs {amount:,}"


def ask_expense() -> dict | None:
    """User se ek kharcha poocho. Khali naam par None."""
    item = input("Kharcha kis cheez ka? ").strip().title()
    if item == "":
        return None
    amount = int(input("Kitne rupay? "))
    category = input("Category? (khana/safar/bill) ").strip().lower()
    return {"item": item, "amount": amount, "category": category}


def total(expenses: list[dict]) -> int:
    """Saare kharchon ka total."""
    return sum(e["amount"] for e in expenses)


def print_report(expenses: list[dict]) -> None:
    """Saare kharche aur total print karo."""
    print("--- Aap ke kharche ---")
    for e in expenses:
        print(f"{e['item']}: {format_rs(e['amount'])}")
    print(f"Total: {format_rs(total(expenses))}")
```

```python
help(format_rs)
```

```python
# program
print("Expense Tracker shuru! (khali Enter = band)")
expenses = []

while True:
    expense = ask_expense()
    if expense is None:
        break
    expenses.append(expense)
    print(f"Add hua: {expense['item']}, {format_rs(expense['amount'])} [{expense['category'].upper()}]")

print_report(expenses)
if total(expenses) > 5000:
    print("Bhai, aaj kharcha zyada ho gaya!")
```
